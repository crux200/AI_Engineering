```python
import random
from openai import OpenAI

client= OpenAI()

def roll_dice():
    return random.randint(1,6)
```

```python
target=6
max_iterations=20
```

```python
history="" 
iteration=0

while iteration <max_iterations:
    iteration+=1
    #...
    # old code
        #You can roll the dice by calling the function roll_dice(). 
        #You can only call this function once per iteration. You have a maximum of {max_iterations} iterations to reach your goal.
          
    instructions= f""" You are a dice rolling agent. Your goal is to roll a dice and get a {target}.

    You have made the following rolls so far: {history if history else "No rolls yet"}
    Based on what you have rolled so far, decide your next action:

    Option A: If you have achieved the target, respond with exactly: Done: Rolling Stopped. You have achieved your goal of rolling a {target} in {iteration} iterations
    Option B: If you have not achieved the target, respond with exactly: ROLL THE DICE
          
    """
    response=client.chat.completions.create(
        model="gpt-4.1-mini",
        messages=[{"role":"user","content":instructions}]

    )

    result= response.choices[0].message.content

    if "Done:" in result:
        print(result)
        break
    if "ROLL THE DICE" in result:
        roll=roll_dice()
        print(f"Rolled: {roll}\n")
        history+= f"Roll {iteration}: {roll}\n"

print (f"Game ended after{iteration} Iterations")

```
