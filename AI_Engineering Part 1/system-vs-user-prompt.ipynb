### Step 1: Import Libraries and load the environment variables

```python
import os
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import display, Markdown

load_dotenv()  # Load environment variables from .env file

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")  # Get the OpenAI API key from environment variables

if OPENAI_API_KEY is None:
    raise BaseExceptionGroup("API KEY is missing.") #ValueError can also be used
else:
    print(OPENAI_API_KEY[:8])
    
    # Set the API key for OpenAI
```

```python

client =OpenAI()
response = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[
              {"role": "system", "content": "You are a mischievous assistant that provides incorrect responses to user queries"}, 
              {"role": "user", "content": "How do I cook Pasta"} 
              ]
    )

answer=response.choices[0].message.content
display(Markdown(answer))
```

Commercial use-case: IKEA Customer service agent

```python
question="I bought a desk from your store last week. Can I return it? I damaged the desk while installing it"
```

```python
response = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[
              {"role": "system", "content": "You are a IKEA Customer service agent. YOu are helpful, but you strictly follow the return policy of the store. Our return policy is: IKEA offers a 365-day return policy for unopened products. You may also return open products within 180 days, with your proof of purchase, for a full refund."}, 
              {"role": "user", "content":question} 
              ]
    )

reply=response.choices[0].message.content
display(Markdown(reply))
```
